```python
# Celda 2: imports y check GPU
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from skimage.feature import hog
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, utils

print("TensorFlow version:", tf.__version__)
gpus = tf.config.list_physical_devices('GPU')
print("GPUs disponibles:", gpus)
```

```python
# Celda 3: cargar dataset MNIST
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
print("Train:", x_train.shape, y_train.shape)
print("Test:", x_test.shape, y_test.shape)

# Mostrar algunos ejemplos
fig, axes = plt.subplots(2,5, figsize=(10,5))
axes = axes.ravel()
for i in range(10):
    axes[i].imshow(x_train[i], cmap='gray')
    axes[i].set_title(f"etiqueta: {y_train[i]}")
    axes[i].axis('off')
plt.show()
```

```python
# Celda 4: funciones de preprocesamiento
def deskew(img):
    # img: numpy 2D uint8
    m = cv2.moments(img)
    if abs(m['mu02']) < 1e-2:
        return img.copy()
    skew = m['mu11'] / m['mu02']
    h, w = img.shape
    M = np.float32([[1, skew, -0.5*skew*w],[0,1,0]])
    img = cv2.warpAffine(img, M, (w, h), flags=cv2.WARP_INVERSE_MAP | cv2.INTER_LINEAR, borderMode=cv2.BORDER_REPLICATE)
    return img

def preprocess_img(img, size=28, blur=1):
    # img: 2D uint8 e.g. MNIST
    # 1) Gaussian blur (opcional)
    if blur>0:
        img = cv2.GaussianBlur(img, (3,3), blur)
    # 2) Otsu binarization
    _, img_th = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    # 3) deskew
    img_ds = deskew(img_th)
    # 4) find bounding box of the digit
    coords = cv2.findNonZero(img_ds)
    if coords is None:
        h, w = img_ds.shape
        cropped = img_ds
    else:
        x, y, w_box, h_box = cv2.boundingRect(coords)
        cropped = img_ds[y:y+h_box, x:x+w_box]
    # 5) resize preserving aspect ratio and pad to size x size
    h_c, w_c = cropped.shape
    if h_c == 0 or w_c == 0:
        resized = cv2.resize(img_ds, (size, size))
    else:
        scale = size / max(h_c, w_c)
        new_w, new_h = int(w_c*scale), int(h_c*scale)
        resized = cv2.resize(cropped, (new_w, new_h), interpolation=cv2.INTER_AREA)
        # pad
        pad_top = (size - new_h) // 2
        pad_bottom = size - new_h - pad_top
        pad_left = (size - new_w) // 2
        pad_right = size - new_w - pad_left
        resized = cv2.copyMakeBorder(resized, pad_top, pad_bottom, pad_left, pad_right, cv2.BORDER_CONSTANT, value=0)
    # 6) normalize to [0,1]
    resized = resized.astype('float32') / 255.0
    return resized

# probar en un ejemplo
example = x_train[0]
proc = preprocess_img(example*1)  # MNIST ya es grayscale 0-255
plt.figure(figsize=(6,3))
plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(example, cmap='gray')
plt.axis('off')
plt.subplot(1,2,2)
plt.title("Preprocesada")
plt.imshow(proc, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# Celda 5: preprocesar todo (train y test). Esto puede tomar varios segundos/minutos.
SIZE = 28

def preprocess_dataset(X):
    Xp = np.zeros((X.shape[0], SIZE, SIZE), dtype=np.float32)
    for i in range(X.shape[0]):
        Xp[i] = preprocess_img(X[i], size=SIZE)
    return Xp

# Para acelerar la demo puedes procesar una porción, pero aquí procesamos todos.
X_train_p = preprocess_dataset(x_train)
X_test_p = preprocess_dataset(x_test)
print("Preprocesado:", X_train_p.shape, X_test_p.shape)
```

```python
# Celda 6: features
# 1) píxeles aplanados
X_train_pix = X_train_p.reshape((X_train_p.shape[0], -1))
X_test_pix = X_test_p.reshape((X_test_p.shape[0], -1))

# 2) HOG
def compute_hog_set(X):
    hog_list = []
    for img in X:
        img_uint8 = (img*255).astype('uint8')
        fd = hog(img_uint8, orientations=9, pixels_per_cell=(8,8),
                 cells_per_block=(2,2), block_norm='L2-Hys', visualize=False)
        hog_list.append(fd)
    return np.array(hog_list)

# Para acelerar puedes usar una muestra; aquí calculamos para todo (toma tiempo)
X_train_hog = compute_hog_set(X_train_p)
X_test_hog = compute_hog_set(X_test_p)
print("Pixel feats:", X_train_pix.shape, "HOG feats:", X_train_hog.shape)
```

```python
# Celda 7: entrenar baselines (usaremos HOG por ser más compacto)

# Para rapidez en Colab, podemos entrenar con una muestra; cambia n_samples=None para todo el set
n_samples = 20000  # ajustar si quieres todo (60000)
if n_samples is None:
    Xb_hog, yb = X_train_hog, y_train
else:
    Xb_hog, yb = X_train_hog[:n_samples], y_train[:n_samples]

# 1) Regresión logística (multinomial)
logreg = LogisticRegression(max_iter=200, solver='saga', multi_class='multinomial', n_jobs=-1)
print("Entrenando LogisticRegression...")
logreg.fit(Xb_hog, yb)
pred_log = logreg.predict(X_test_hog)
print("Acc LogisticRegression:", accuracy_score(y_test, pred_log))

# 2) KNN
knn = KNeighborsClassifier(n_neighbors=3, n_jobs=-1)
print("Entrenando KNN...")
knn.fit(Xb_hog, yb)
pred_knn = knn.predict(X_test_hog)
print("Acc KNN:", accuracy_score(y_test, pred_knn))

# 3) Linear SVM
svc = LinearSVC(max_iter=5000)
print("Entrenando LinearSVC...")
svc.fit(Xb_hog, yb)
pred_svc = svc.predict(X_test_hog)
print("Acc LinearSVC:", accuracy_score(y_test, pred_svc))

# Mostrar reporte para el mejor (ejemplo: logreg)
print("\nClassification report (LogisticRegression):\n")
print(classification_report(y_test, pred_log, digits=4))
```

```python
# Celda 8: clustering con KMeans (10 clusters)
kmeans = KMeans(n_clusters=10, random_state=42, n_init=10)
print("Entrenando KMeans sobre HOG (muestra para rapidez)...")
kmeans.fit(Xb_hog)
labels_km = kmeans.predict(X_test_hog)

# Mapear cluster a etiqueta mayoritaria para evaluar "purity"
from collections import defaultdict, Counter
cluster_to_label = defaultdict(list)
for cluster, true in zip(kmeans.labels_, yb):
    cluster_to_label[cluster].append(true)

cluster_map = {}
for c, lst in cluster_to_label.items():
    cluster_map[c] = Counter(lst).most_common(1)[0][0]

mapped = [cluster_map.get(c, -1) for c in labels_km]
print("Acc (KMeans mapeado a etiquetas):", accuracy_score(y_test, mapped))
```

```python
# Celda 9: definir CNN en Keras
input_shape = (SIZE, SIZE, 1)
num_classes = 10

def build_cnn():
    model = models.Sequential([
        layers.Input(shape=input_shape),
        layers.Conv2D(32, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(64, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(128, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

cnn = build_cnn()
cnn.summary()
```

```python
# Celda 10: preparar datos y data augmentation
X_train_cnn = X_train_p[..., np.newaxis]  # shape (n,28,28,1)
X_test_cnn = X_test_p[..., np.newaxis]

# normalización: ya está en [0,1]
# Data augmentation con ImageDataGenerator
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rotation_range=12,
    width_shift_range=0.08,
    height_shift_range=0.08,
    shear_range=0.08,
    zoom_range=0.08
)
datagen.fit(X_train_cnn)

# Callbacks
es = callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
rlr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3)
```

```python
# CELDA: Entrenar CNN (Opción A: split manual y validation_data)
from sklearn.model_selection import train_test_split

# dividir manualmente (mantener estratificación)
X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train_cnn, y_train, test_size=0.10, random_state=42, stratify=y_train)

BATCH = 128
EPOCHS = 15

train_gen = datagen.flow(X_train_sub, y_train_sub, batch_size=BATCH)

steps_per_epoch = max(1, len(X_train_sub) // BATCH)
validation_steps = max(1, len(X_val) // BATCH)

history = cnn.fit(
    train_gen,
    epochs=EPOCHS,
    steps_per_epoch=steps_per_epoch,
    validation_data=(X_val, y_val),
    callbacks=[es, rlr]
)
```

```python
# CELDA: Visualizar curvas de entrenamiento
import matplotlib.pyplot as plt

# history viene de la celda de entrenamiento
hist = history.history

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(hist['loss'], label='train_loss')
plt.plot(hist.get('val_loss', []), label='val_loss')
plt.title('Pérdida durante entrenamiento')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1,2,2)
plt.plot(hist['accuracy'], label='train_acc')
plt.plot(hist.get('val_accuracy', []), label='val_acc')
plt.title('Accuracy durante entrenamiento')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

```python
# CELDA: Evaluación en test y métricas
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
import seaborn as sns
import numpy as np

# Evaluación rápida de Keras
test_loss, test_acc = cnn.evaluate(X_test_cnn, y_test, verbose=0)
print(f"CNN Test loss: {test_loss:.4f}  |  Test accuracy: {test_acc:.4f}")

# Predicciones
y_prob = cnn.predict(X_test_cnn)
y_pred = np.argmax(y_prob, axis=1)

# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Matriz de Confusión (CNN)")
plt.ylabel("Etiqueta real")
plt.xlabel("Predicción")
plt.show()

# Reporte por clase
print("Classification Report:\n")
print(classification_report(y_test, y_pred, digits=4))
```

```python
# CELDA: Mostrar ejemplos correctos e incorrectos
import numpy as np
import matplotlib.pyplot as plt

def show_examples(X, y_true, y_pred, n=12):
    idxs = np.arange(len(y_true))
    mis = idxs[y_true != y_pred]
    cor = idxs[y_true == y_pred]
    sel = np.concatenate([mis[:n//2], cor[:n//2]])
    plt.figure(figsize=(12,6))
    for i, idx in enumerate(sel):
        plt.subplot(3,4,i+1)
        plt.imshow(X[idx].squeeze(), cmap='gray')
        plt.title(f"V:{y_true[idx]} P:{y_pred[idx]}")
        plt.axis('off')
    plt.show()

show_examples(X_test_cnn, y_test, y_pred, n=12)
```
